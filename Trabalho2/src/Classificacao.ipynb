```python
# data preparation

# target value
y = data.loc[:, "utilizRede"].values

# feature variable
X = data.drop("utilizRede", axis=1).values
```

```python
label_encoder = LabelEncoder()

# fit and transform the target variable
y = label_encoder.fit_transform(y)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

1) Usando o método $\boldsymbol{k}$fold cross validation desenvolva modelos de previsão de utilizRede usando os seguintes métodos:

a) Árvore de Decisão

```python
# Decision Tree with Max Depth = 3

tree_model = DecisionTreeClassifier(random_state=42, max_depth=3)

# Apply K-fold Cross Validation

metrics = ['accuracy', 'precision_macro', 'recall_macro', 'f1_macro']

scores = {}
for metric in metrics:
    scores[metric] = cross_val_score(tree_model, X, y, cv=5, scoring=metric).mean()

# Results (Mean and Standard Deviation)
for metric in metrics:
    mean_val = scores[metric]
    std_val = cross_val_score(tree_model, X, y, cv=5, scoring=metric).std()
    print(f"{metric}: {mean_val:.3f} (+/- {std_val:.3f})")
```

```python
# build the model on the training data
tree_model.fit(X_train, y_train)

plt.style.use('ggplot')

# plot tree
plt.figure(figsize=(12, 15))  # set plot size

plot_tree(tree_model,
          # Names of the input features
          feature_names=list(data.columns),
          # Names of the classes
          class_names=list(data["utilizRede"]),
          # Color Nodes by Purity
          filled=True,
          # Show percentage of samples instead of numbers
          proportion=True,
          # acute rounded nodes
          node_ids=True
)
plt.show()
```

```python
#
tree_model.score(X_test, y_test)
tree_model.score(X_train, y_train)

# Make Predictions
y_train_pred = tree_model.predict(X_train)
y_test_pred = tree_model.predict(X_test)

# Show Results
tree_train = accuracy_score(y_train, y_train_pred)
print("Accuracy on training set:", tree_train)

tree_test = accuracy_score(y_test, y_test_pred)
print("Accuracy on test set:", tree_test)
```

```python
# Confusion Matrix - Training Set
c_matrix_train = confusion_matrix(y_train, tree_model.predict(X_train))

disp = ConfusionMatrixDisplay(
    confusion_matrix= c_matrix_train,
    display_labels= ['baixo', 'medio', 'alto']
)

print('Confusion Matrix (Train Set)')
disp.plot()
```

```python
# Confusion Matrix - Test Set
c_matrix_test = confusion_matrix(y_test, tree_model.predict(X_test))

disp = ConfusionMatrixDisplay(
    confusion_matrix= c_matrix_test,
    display_labels= ['baixo', 'medio', 'alto']
)

print('Confusion Matrix (Test Set)')
disp.plot()
```

```python
# Creating Object
scaler = StandardScaler()

# Standardizing the features
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

# Creating (MLP) classifier
nn = MLPClassifier(hidden_layer_sizes=1, max_iter=1000, random_state=42)

# Train the Model
nn.fit(X_train, y_train)

# Making Prediction
y_pred = nn.predict(X_test)

# Determining Accuracy
accuracy = accuracy_score(y_test, y_pred)
print(f"Accuracy: {accuracy * 100:.2f}%")

# Confusion Matrix

cm = confusion_matrix(y_test, y_pred)

print("----------------- Classification Report ------------------")
print(classification_report(y_test, y_pred, zero_division=0))

disp = ConfusionMatrixDisplay(confusion_matrix=cm)

disp.plot(cmap='viridis')

plt.show()
```

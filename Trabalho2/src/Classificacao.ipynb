### Configuração #1 - Uma hidden layer com 10 neuronios

```python
# Defining Metrics
scoring = ['accuracy', 'precision_macro', 'recall_macro', 'f1_macro']

# Creating Object
scaler = StandardScaler()

# Standardizing the features
X_train_normalized = scaler.fit_transform(X_train)
X_test_normalized = scaler.transform(X_test)
```

```python
# Creating (MLP) classifier
nn = MLPClassifier(
    hidden_layer_sizes=(10,),  # Number of hidden Layers (1 layer with 10 neurons)
    max_iter=1000,  # Number of Iterations
    random_state=42,  # Random State
    validation_fraction=0.1,  # Fraction of Training Data to be used as Validation Set
    early_stopping=True,  # Early Stopping
    learning_rate_init=0.001  # Learning Rate
)

# Train the Model
nn.fit(X_train, y_train)

# Making Prediction
y_pred = nn.predict(X_test)

print("----------------- Classification Report ------------------")
print(classification_report(y_test, y_pred, zero_division=0))
```

```python
nn_config_01_results = cross_validate(nn, X_train_normalized, y_train, cv=10, scoring=scoring)

# Determining Accuracy
accuracy = accuracy_score(y_test, y_pred)
print(f"Accuracy: {accuracy * 100:.2f}%\n")

# Iteration where Model stopped improving
n_iterations = nn.n_iter_
print(f"O modelo parou de melhorar na iteração: {n_iterations}\n")

print(f"Métricas Médias (K-fold):")
for metric in scoring:
    mean_val = nn_config_01_results[f'test_{metric}'].mean()
    std_val = nn_config_01_results[f'test_{metric}'].std()
    print(f"{metric.split('_')}: {mean_val:.3f} (+/- {std_val:.3f})")
```

(3) discussão da influência da learning rate e do early stopping.

### Influência da Learning Rate

A learning rate define a magnitude do ajuste dos pesos em cada interação.

- Se for muito alto, o modelo pode aprender muito rápido e "saltar" o ponto de erro minímo, causando variações bruscas na curva de loss ou mesmo impedindo a convergência.

- Se for muito baixa, o treino torna-se excessovamente lento, podendo ficar preso em minímos locais antes de atingir o desempenho ideal.

---

### Influência do Early Stopping

Este é um mecanismo de regularização que interrompe o treino assim que o desempenho nos dados de validação deixa de melhorar.

- Sem o early stopping, a rede pode continuar a ajustar-se aos dados de treino até os memorizar (overfitting), o que prejudica a capacidade de lidar com dados novos.
- Evita o desperdício de recursos computacionais ao não completar as iterações máximas se o modelo já não estiver a evoluir.

d. K -vizinhos-mais-próximos. Optimize o parâmetro K.

```python
# define interval for k (only even)
k_values = range(1, 22, 2)

mean_accuracy = []

for k in k_values:
    # Create KNeighbors Model
    knn = KNeighborsClassifier(n_neighbors=k)

    # Apply K-fold Cross Validation
    K_neighbors_scores = cross_validate(knn, X_sample_normalized, y_sample, cv=10, scoring=scoring)

    mean_accuracy.append(K_neighbors_scores['test_accuracy'].mean())

    # Results
    print("Metrics for k = ", k)
    print(f"Accuracy:  {K_neighbors_scores['test_accuracy'].mean():.3f} (+/- {K_neighbors_scores['test_accuracy'].std():.3f})")
    print(f"Precision_Macro-Score:  {K_neighbors_scores['test_precision_macro'].mean():.3f}")
    print(f"Recall_Macro-Score:  {K_neighbors_scores['test_recall_macro'].mean():.3f}")
    print(f"F1-Score:  {K_neighbors_scores['test_f1_macro'].mean():.3f}")

    print("--------------------------------------------------------------------")
```

```python
# Visualize graphic to find the "great point"
plt.figure(figsize=(10, 5))
plt.plot(k_values, mean_accuracy, marker='o', linestyle='dashed', color='green')
plt.title('Acurácia vs. Valor de K')
plt.xlabel('Valor de K')
plt.ylabel('Acurácia Média')
plt.grid(True)
plt.show()
```

```python
# Identify the best k
best_k = list(k_values)[np.argmax(mean_accuracy)]

print(f"O melhor valor de K identificado é: {best_k}")
```

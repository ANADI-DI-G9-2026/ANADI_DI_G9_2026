1) Usando o método $\boldsymbol{k}$fold cross validation desenvolva modelos de previsão de utilizRede usando os seguintes métodos:

a) Árvore de Decisão

```python
# Decision Tree with Max Depth = 3

tree_model = DecisionTreeClassifier(random_state=42, max_depth=3)

# Apply K-fold Cross Validation

metrics = ['accuracy', 'precision_macro', 'recall_macro', 'f1_macro']

scores = {}
for metric in metrics:
    scores[metric] = cross_val_score(tree_model, X, y, cv=5, scoring=metric).mean()

# Results (Mean and Standard Deviation)
for metric in metrics:
    mean_val = scores[metric]
    std_val = cross_val_score(tree_model, X, y, cv=5, scoring=metric).std()
    print(f"{metric}: {mean_val:.3f} (+/- {std_val:.3f})")
```

```python
# build the model on the training data
tree_model.fit(X_train, y_train)

plt.style.use('ggplot')

# plot tree
plt.figure(figsize=(12, 15))  # set plot size

plot_tree(tree_model,
          # Names of the input features
          feature_names=list(data.columns),
          # Names of the classes
          class_names=list(data["utilizRede"]),
          # Color Nodes by Purity
          filled=True,
          # Show percentage of samples instead of numbers
          proportion=True,
          # acute rounded nodes
          node_ids=True
)
plt.show()
```

```python
# Confusion Matrix - Training Set
c_matrix_train = confusion_matrix(y_train, tree_model.predict(X_train))

disp = ConfusionMatrixDisplay(
    confusion_matrix= c_matrix_train,
    display_labels= ['baixo', 'medio', 'alto']
)

print('Confusion Matrix (Train Set)')
disp.plot()
```

```python
# Confusion Matrix - Test Set
c_matrix_test = confusion_matrix(y_test, tree_model.predict(X_test))

disp = ConfusionMatrixDisplay(
    confusion_matrix= c_matrix_test,
    display_labels= ['baixo', 'medio', 'alto']
)

print('Confusion Matrix (Test Set)')
disp.plot()
```

### Configuração #1 - Uma hidden layer com 10 neuronios

```python
# Creating Object
scaler = StandardScaler()

# Standardizing the features
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

# Creating (MLP) classifier
nn = MLPClassifier(
    hidden_layer_sizes=(10, ), # Number of hidden Layers (1 layer with 10 neurons
    max_iter=1000,             # Number of Iterations
    random_state=42,           # Random State
    validation_fraction=0.1,   # Fraction of Training Data to be used as Validation Set
    early_stopping=True,       # Early Stopping
    learning_rate_init=0.001   # Learning Rate
)

# Train the Model
nn.fit(X_train, y_train)

# Making Prediction
y_pred = nn.predict(X_test)

# Determining Accuracy
accuracy = accuracy_score(y_test, y_pred)
print(f"Accuracy: {accuracy * 100:.2f}%")

print("----------------- Classification Report ------------------")
print(classification_report(y_test, y_pred, zero_division=0))
```

```python
# Confusion Matrix
cm = confusion_matrix(y_test, y_pred)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels= ['baixo', 'medio', 'alto']
)
disp.plot(cmap='viridis')

plt.show()
```

```python
# Curva de Loss
plt.figure(figsize=(10, 5))

# Loss Curve
plt.plot(nn.loss_curve_, label='Treino', color='blue')

# Validation Curve
plt.plot(nn.validation_scores_, label='Validação (Acurácia)', color='red', linestyle='--')

plt.title('Curvas de Aprendizagem: Treino vs Validação')

plt.xlabel('Iterações')
plt.ylabel('Valor')

plt.legend()
plt.grid(True)

plt.show()
```

c. SVM. Optimize o kernel.

```python
data_sample = data.sample(n=10000, random_state=42)

X_sample = data_sample.drop('utilizRede', axis=1)
y_sample = data_sample['utilizRede']

scaler = StandardScaler()

X_sample_normalized = scaler.fit_transform(X_sample)
```

```python
scoring = ['accuracy', 'precision_macro', 'recall_macro', 'f1_macro']
```

```python
# Kernels for testing
kernels = ['linear', 'poly', 'rbf', 'sigmoid']

for k in kernels:
    # Create Model with Specific kernel
    svm_model = SVC(
        kernel=k,
        random_state=42
    )

    # Execute Cross Validation
    scores = cross_validate(svm_model, X_sample_normalized, y_sample, cv=5, scoring=scoring)

    print(f"\n--- Kernel: {k.upper()} ---")
    print(f"Accuracy:  {scores['test_accuracy'].mean():.3f} (+/- {scores['test_accuracy'].std():.3f})")
    print(f"F1-Score:  {scores['test_f1_macro'].mean():.3f}")
```

d. K -vizinhos-mais-próximos. Optimize o parâmetro K.

```python
# define interval for k (only even)
k_values = range(1, 22, 2)

mean_accuracy = []

for k in k_values:
    knn = KNeighborsClassifier(n_neighbors=k)
    scores = cross_validate(knn, X_sample_normalized, y_sample, cv=5, scoring=scoring)
    mean_accuracy.append(scores['test_accuracy'].mean())
```

```python
# Visualize graphic to find the "great point"
plt.figure(figsize=(10, 5))
plt.plot(k_values, mean_accuracy, marker='o', linestyle='dashed', color='green')
plt.title('Acurácia vs. Valor de K')
plt.xlabel('Valor de K')
plt.ylabel('Acurácia Média')
plt.grid(True)
plt.show()
```

```python
# Identify the best k
best_k = list(k_values)[np.argmax(mean_accuracy)]

print(f"O melhor valor de K identificado é: {best_k}")
```
